# Using scikit learn to encode categorical features(One-hot encoding)

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
df_full_train[['gender', 'contract']].iloc[:10]
```

```python
train_dicts = df_train[categorical + numerical].to_dict(orient='records')
train_dicts[0].keys()
```

```python
dv = DictVectorizer(sparse = False)
dv
```

```python
dv.fit(train_dicts)
```

```python
dv.feature_names_
```

```python
X_train = dv.transform(train_dicts)
X_train
#Create sparse matrix. Since we do not want sparse matrix and want full numpy arrays, we will set sparse to False in dv.
```

```python
X_train.shape
```

```python
val_dicts = df_val[numerical + categorical].to_dict(orient = 'records')
```

```python
X_val = dv.transform(val_dicts)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter = 300)
```

```python
model.fit(X_train, y_train)
```

```python
model.intercept_
#intercept is synonym for bias or w0
```

```python
model.coef_.round(3)
```

```python
y_pred =model.predict_proba(X_val)[:, 1]
```

```python
churn_decision= (y_pred >= 0.5)
```

```python
df_val[churn_decision]
```
